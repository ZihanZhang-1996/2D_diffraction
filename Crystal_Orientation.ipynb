```python
from ase.visualize import view
from ase.build import bulk
from ase.io import read
from ase.io import write
from ase.io import cif
import os
```

### This program takes the .cif file and output 3 .cif with 3 different orientation,
#### $\vec{a}, \vec{b}, \vec{c}$
#### $\vec{a}', \vec{b}', \vec{c}'$ = $\vec{c}, \vec{a}, \vec{b}$
#### $\vec{a}'', \vec{b}'', \vec{c}''$ = $\vec{b}, \vec{c}, \vec{a}.$

```python
path = '../Huilong/'
filename = 'new.cif'
filepath = os.path.join(path, filename)

atoms1 = read(filepath)

output_file = os.path.join(path, 'Orientation1.cif')  
atoms1.write(output_file)
```

```python
atoms1 = read(filepath)
old_cell = atoms1.get_cell()
old_frac = atoms1.get_scaled_positions()

# Step 1: Permute lattice vectors: a' = c, b' = a, c' = b
new_cell = [old_cell[2], old_cell[0], old_cell[1]]
atoms1.set_cell(new_cell, scale_atoms=False)

# Step 2: Permute fractional coordinates: x' = z, y' = x, z' = y
new_frac = old_frac[:, [2, 0, 1]]
atoms1.set_scaled_positions(new_frac)
output_file = os.path.join(path, 'Orientation2.cif')  
atoms1.write(output_file)
```

```python
atoms1 = read(filepath)
old_cell = atoms1.get_cell()
old_frac = atoms1.get_scaled_positions()

# Step 2: Apply cell permutation: a' = b, b' = c, c' = a
new_cell = [old_cell[1], old_cell[2], old_cell[0]]
atoms1.set_cell(new_cell, scale_atoms=False)

# Step 3: Permute fractional coordinates accordingly: x' = y, y' = z, z' = x
new_frac = old_frac[:, [1, 2, 0]]
atoms1.set_scaled_positions(new_frac)
output_file = os.path.join(path, 'Orientation3.cif')  
atoms1.write(output_file)
```
